## Requêtes POST

Source : [Datacamp_Request](https://www.datacamp.com/fr/tutorial/making-http-requests-in-python)

````python
# Define new data to create
new_data = {
    "userID": 1,
    "id": 1,
    "title": "Making a POST request",
    "body": "This is the data we created."
}

# The API endpoint to communicate with
url_post = "https://jsonplaceholder.typicode.com/posts"

# A POST request to tthe API
post_response = requests.post(url_post, json=new_data)

# Print the response
post_response_json = post_response.json()
print(post_response_json)
````

## Questions

**Quelle est la différence entre r.text , r.content et r.json()**


`r.text` permet de récupérer le cors de la réponse sous forme de texte.

`r.content` permet de récupérer les données brutes (sous forme d'octets).

`r.json()` décode le JSON et le convertit en dictionnaire python.

**Quelle est la différence entre data= et json= dans un POST ?**

`json=` permet d'envoyer des données sous forme d'un JSON lors d'une requête POST.<br>
(requests convertit un dictionnaire python en JSON)

```py
payload = {
    "name": "Alice",
    "age": 25
}

r = requests.post(url, json=payload)
```

`data=` permet d'envoyer des données de formulaires ou des données brutes lors d'une requête POST<br>

```py
payload = {
    "username": "Alice",
    "password": "12345"
}

r = requests.post(url, data=payload)
```

Requests convertir les données au format : `username=Alice&password=12345`
